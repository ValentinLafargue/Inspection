# Saving and loading through HuggingFace

```python
HUGGINGFACEHUB_API_TOKEN = ""
login(HUGGINGFACEHUB_API_TOKEN)
```

```python
repo_id = "ValentinLAFARGUE/EIF-biased-classifier"
dic_X_train_dim = {
    'INC' : 8,
    'TRA' : 19,
    'MOB' : 19,
    'EMP' : 15,
    'PUC' : 18,
    'ADULT' : 34,
    'BAF': 42,
}
```

## ASC models

```python
for file_obj in ['INC', 'TRA', 'MOB', 'EMP', 'PUC']:

    
    subfolder = "ASC_" + file_obj + "_model"

    result_path = '../Result_2/ASC_' + file_obj + '/'
    net_path = result_path + 'net_state_dic.pt'

    model = Network(
        dic_X_train_dim[file_obj],
        activation_bool=True,
        n_nodes=256,
        n_loop=2
    )

    model.load_state_dict(torch.load(net_path, weights_only=True))

    # save locally
    model.save_pretrained(subfolder)

    # upload to repo/subfolder
    api = HfApi()
    api.upload_folder(
        folder_path=subfolder,
        repo_id=repo_id,
        path_in_repo=subfolder
    )

    '''
    # reload
    model = Network.from_pretrained(
        repo_id,
        subfolder=subfolder
    )
    '''
```

## Adult

```python
file_obj = 'ADULT' 
subfolder = file_obj + "_model"

result_path = '../Result_2/' + file_obj + '/'
net_path = result_path + 'net_state_dic.pt'

model = Network(
    dic_X_train_dim[file_obj],
    activation_bool=True,
    n_nodes=256,
    n_loop=2
)

model.load_state_dict(torch.load(net_path, weights_only=True))

# save locally
model.save_pretrained(subfolder)

# upload to repo/subfolder
api = HfApi()
api.upload_folder(
    folder_path=subfolder,
    repo_id=repo_id,
    path_in_repo=subfolder
)

'''
# reload
model = Network.from_pretrained(
    repo_id,
    subfolder=subfolder
)
'''
```

## BAF

```python
file_obj = 'BAF' 
subfolder = file_obj + "_model"

result_path = '../Result_2/' + file_obj + '/'
net_path = result_path + 'net_state_dic.pt'

model = Network(
    dic_X_train_dim[file_obj],
    activation_bool=True,
    n_nodes=1024,
    n_loop=2
)

model.load_state_dict(torch.load(net_path, weights_only=True))

# save locally
model.save_pretrained(subfolder)

# upload to repo/subfolder
api = HfApi()
api.upload_folder(
    folder_path=subfolder,
    repo_id=repo_id,
    path_in_repo=subfolder
)

''' 
model = Network.from_pretrained(
    repo_id,
    subfolder=subfolder
)
'''
```
